# 02 MARBERTv2 Fine-Tuning Workflow

This notebook records training configuration and exports reproducible mock metrics/checkpoint metadata.

```python
from pathlib import Path
import json

root = Path.cwd().resolve().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
out_dir = root / 'notebooks' / 'artifacts'
out_dir.mkdir(parents=True, exist_ok=True)

training_config = {
    'model': 'UBC-NLP/MARBERTv2',
    'max_length': 128,
    'batch_size': 16,
    'epochs': 5,
    'learning_rate': 2e-5,
    'optimizer': 'AdamW',
    'metric_for_best_model': 'f1'
}

metrics = {'accuracy': 0.82, 'precision': 0.80, 'recall': 0.78, 'f1': 0.79}
(out_dir / 'marbert_config.json').write_text(json.dumps(training_config, indent=2), encoding='utf-8')
(out_dir / 'marbert_metrics.json').write_text(json.dumps(metrics, indent=2), encoding='utf-8')
(out_dir / 'marbert_best_checkpoint.txt').write_text('checkpoint-best-marbert', encoding='utf-8')
metrics
```
