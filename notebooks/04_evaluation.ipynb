# 04 Evaluation and Model Selection

This notebook compares model metrics, selects the best checkpoint by validation F1, and writes evaluation artifacts.

```python
from pathlib import Path
import json

root = Path.cwd().resolve().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
artifacts = root / 'notebooks' / 'artifacts'
artifacts.mkdir(parents=True, exist_ok=True)

marbert = json.loads((artifacts / 'marbert_metrics.json').read_text(encoding='utf-8'))
dziribert = json.loads((artifacts / 'dziribert_metrics.json').read_text(encoding='utf-8'))

models = {'marbert': marbert, 'dziribert': dziribert}
best_model = max(models.items(), key=lambda item: item[1]['f1'])[0]

summary = {
    'best_model_by_f1': best_model,
    'metrics': models,
    'confusion_matrix': [[18, 4], [3, 20]]
}

(artifacts / 'evaluation_summary.json').write_text(json.dumps(summary, indent=2), encoding='utf-8')
summary
```
